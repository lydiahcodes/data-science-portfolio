```python
import requests
import pandas as pd

```

```python
url= "https://api.open-meteo.com/v1/forecast"
print("open-meteo api endpoint defined successfully!")
```

```python
query_parameters = {
    "longitude":  36.0800,
    "latitude": -0.3031,
    "hourly": "temperature_2m,relativehumidity_2m",
    "forecast_days" : 1
}
print (query_parameters)
```

```python
print("sending API request to open-meteo api endpoint...")
server_response = requests.get(url, params=query_parameters)
print("request completed")
```

```python
print("HTTP status code:", server_response.status_code)
```

```python
if server_response.status_code == 200:
  print("connection is a success")
  json_data = server_response.json()
  print(json_data)
else:
  print("request failed:", server_response.status_code)
```

```python
hourly_dictionary = json_data["hourly"]
print(hourly_dictionary.keys())
```

```python
weather_df = pd.DataFrame({
    "date_time": pd.to_datetime(hourly_dictionary["time"]),
    "temperature_celcius": hourly_dictionary["temperature_2m"],
    "relative_humidity": hourly_dictionary["relativehumidity_2m"]
})
print(weather_df.head())
```

```python
print("dataset dimensions:", weather_df.shape)
print("\nmissing values:")
print(weather_df.isnull().sum())
```

```python
weather_df.to_csv("weather_data.csv", index=False)
print("dataset saved to weather_data.csv")
```
